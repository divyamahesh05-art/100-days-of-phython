```python
import pandas as pd

# Table 1: Patient Demographics (Contains extra spaces & bad casing)
patient_data = {
    "Patient_ID": ["P101", "P102", "P103", "P104"],
    "Patient_Name": ["  dr. sara ali ", "RAVI KUMAR", "john smith  ", "  FATIMA HASSAN "]
}

# Table 2: Lab Results (Contains unit text)
lab_data = {
    "Patient_ID": ["P101", "P102", "P103"],
    "Test_Name": ["FBS (Fasting)", "lipid profile", "FBS (Fasting)"],
    "Result": ["135 mg/dL", "210 mg/dL", "98 mg/dL"]
}

df_patients = pd.DataFrame(patient_data)
df_lab = pd.DataFrame(lab_data)
```

```python
df_patients["Clean_Name"] = df_patients["Patient_Name"].str.strip().str.title()
```

```python
df_lab["Result"] = df_lab["Result"].str.replace(" mg/dL", "")
```

```python
df_merged = pd.merge(df_patients, df_lab, on="Patient_ID", how="left")
```

```python
fbs_patient = df_merged[df_merged["Test_Name"].str.contains("fasting", case=False, na=False)]

print(fbs_patient[["Patient_ID", "Clean_Name", "Test_Name", "Result"]])
```
