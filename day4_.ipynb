```python
while True:
    temp = float(input("Enter temperature: "))

    if temp > 102.0:
        print("High Fever!")
    elif temp > 98.6:
        print("Mild Fever.")
    else:
        print("Normal Temperature! Stopping monitor...")
        break  # ഇവിടെ ലൂപ്പ് പെട്ടെന്ന് നിൽക്കും!
```

```python
reagent_stock = 50
while reagent_stock > 0:
  used = int(input("Enter reagent used for test (in ml): "))
  reagent_stock = reagent_stock - used
  if reagent_stock > 10:
    print(f"Reagent remaining: {reagent_stock} ml")
  elif reagent_stock > 0 and reagent_stock <= 10:
    print(f"WARNING: Low stock! Only {reagent_stock} ml remaining!")
  else:
    print(f"WARNING: Low stock! Only {reagent_stock} ml remaining!")
```

```python
normal_reading_count = 0
while True:
  bpm = int(input("Enter Heart Rate (BPM) [or enter 0 to stop]: "))
  if bpm == 0:
    print ("Monitoring stopped manually")
    break
  elif bpm < 60 or bpm > 100:
    print("ALERT: Abnormal Heart Rate!")
  else:
    print("Heart rate is Normal.")
  normal_reading_count = normal_reading_count + 1
  if normal_reading_count == 3:
    print("Patient is stable for 3 consecutive readings. Discontinuing monitor.")
    break


```

```python
successful_tests = 0
rejected_tests = 0
while True:
  hb = float(input("Enter Hemoglobin level (g/dL) [or 0 to stop]: "))
  if hb == 0:
    print("Manual shutdown initiated.")
    break
  elif hb < 12.0 or hb > 17.5:
    print("Sample Rejected: Hemoglobin value out of valid range!")
    rejected_tests = rejected_tests + 1
  else:
    print("Sample Accepted: Hemoglobin value within valid range.")
    successful_tests = successful_tests + 1
    if successful_tests == 3:
      print("Daily quota of 3 successful samples reached!")
    elif rejected_tests == 2:
      print("ALERT: 2 samples rejected! Machine auto-shutdown for calibration!")
      break
    print(f"Total Approved: {successful_tests} | Total Rejected: {rejected_tests}")
```
