```python
import json
```

```python
dataset = pd.read_csv('employees.csv')
```

```python
mean_salary = round(dataset['Salary'].mean(),2)
std_salary= round(dataset['Salary'].std(),2)
```

```python
filtered_data=dataset[(dataset['Salary']>=50000)&(dataset['Salary']<=70000)]
prob=round(len(filtered_data)/len(dataset),2)
```

```python
print(f"Mean Salary: {mean_salary}")
print(f"Standard Deviation of Salary: {std_salary}")
print(f"Probability of Salary between 50000 and 70000: {prob}")
```

```python
submission = {
    "2.1": mean_salary,
    "2.2": std_salary,
    "2.3": prob
}
with open('submission.json','w')as file:
  json.dump(submission,file,indent=4)
print("json created")
```

```python
import numpy as np
```

```python
patient_data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105"],
    "Name": ["Sarah", "Raj", "John", "Meera", "Anil"],
    "Age": [32, np.nan, 45, 70, 28],
    "Hb_Level": [11.2, 14.5, np.nan, 13.0, 10.1],
    "FBS": [135, 95, 160, np.nan, 105]
}
```

```python
df = pd.DataFrame(patient_data)
```

```python
print("--- PATIENT DATA WITH MISSING VALUES ---")
print(df)
```

```python
print(df.isnull().sum())
```

```python
cleane_df=df.dropna()
```

```python
print("\n--- CLEANED DATA (AFTER REMOVING MISSING VALUES) ---")
print(cleane_df)
```

```python
mean_Hb=df['Hb_Level'].mean()
```

```python
df["Hb_Level"] = df["Hb_Level"].fillna(mean_Hb)
```

```python
print(df)
```

```python
mean_FBS=df['FBS'].mean()
df['FBS']=df['FBS'].fillna(mean_FBS)
print(df)
```

```python
lab_data = {
    "Patient": ["Amina", "Ravi", "John", "Fatima"],
    "Age": [25, 62, np.nan, 45],
    "Hb_Level": [12.5, 9.8, 11.0, np.nan]}
```

```python
df=pd.DataFrame(lab_data)
print(df)
```

```python
mean_Hb=df['Hb_Level'].mean()
```

```python
df['Hb_Level']=df['Hb_Level'].fillna(mean_Hb)
```

```python
mean_age=df['Age'].mean()
df['Age']=df['Age'].fillna(mean_age)
df = df.rename(columns={"Patient": "Patient_Name"})
print(df)
```

```python
anemic_patients = df[df["Hb_Level"] < 11.0]
print(anemic_patients)
```

```python
data = {
    "Patient_Name": ["Amina", "Ravi", "John", "Fatima", "Anil", "Sara"],
    "Gender": ["Female", "Male", "Male", "Female", "Male", "Female"],
    "Department": ["OP", "IP", "IP", "OP", "OP", "IP"],
    "Hb_Level": [12.5, 9.8, 11.0, 10.2, 14.0, 11.8],
    "Bill_Amount": [500, 1200, 1500, 600, 450, 1100]
}

df = pd.DataFrame(data)
```

```python
print("--- PATIENT DATA ---")
print(df)
```

```python
gender_Hb=df.groupby('Gender')['Hb_Level'].mean()
print("\n--- AVERAGE HB LEVEL BY GENDER ---")
print(gender_Hb)
```

```python
deparment_bill=df.groupby('Department')['Bill_Amount'].sum()
print("\n--- TOTAL BILL AMOUNT BY DEPARTMENT ---")
print(deparment_bill)
```

```python
multi_group = df.groupby(["Gender", "Department"])["Hb_Level"].mean()

print("--- AVERAGE HB BY GENDER & DEPARTMENT ---")
print(multi_group)
```
