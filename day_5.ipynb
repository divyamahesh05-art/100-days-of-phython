```python
def glucose_check(glucose):
  if glucose <70.0:
    return("low")
  elif glucose >= 110.0 and glucose <=140:
    return("pre diabetic")
  elif glucose >= 70.0 and glucose <= 110.0:
    return("normal")
  else:
    return("diabetic")
```

```python
glucose_sample=[75,110,125,145,180,200]
for glucose in glucose_sample:
  result=glucose_check(glucose)
print(f"Glucose Level: {glucose} mg/dL -> Status: {result}")
```

```python
glucose_samples = [65, 92, 115, 140, 88]
print("--- LAB BATCH REPORT ---")
for sample in glucose_samples:
    status = glucose_check(sample)
    print(f"Glucose Level: {sample} mg/dL -> Status: {status}")
```

```python
def check_hb_status(gender, hb):

  if gender == "Male" and hb < 13.0:

    return "anemic"
  elif gender == "Female" and hb < 12.0:

    return "anemic"
  else:
   return "normal"
```

```python
patients = [("Female", 10.5), ("Male", 14.2), ("Female", 13.0), ("Male", 12.0)]
for patient in patients:
    gender, hb = patient
    status = check_hb_status(gender, hb)
    print(f"Gender: {gender}, HB Value: {hb} -> Status: {status}")
```

```python
def classify_glucose(blood_sugar):
  if blood_sugar < 70:
    return "low"
  elif 70 <= blood_sugar <= 110:
    return "normal"
  else:
    return "high"
```

```python
all_samples = []
while True:
  bs = float(input("Enter Blood Sugar level (mg/dL) [or 0 to finish]: "))
  if bs == 0:
    break
    status = classify_glucose(bs)
  print(f"Blood Sugar Level: {bs} mg/dL -> Status: {status}")
  all_samples.append(bs)
  print("\n--- LAB BATCH SUMMARY REPORT ---")
print("Total Samples Processed:", len(all_samples))
print("Highest Glucose Level:", max(all_samples))
print("Lowest Glucose Level:", min(all_samples))
print("Average Glucose Level:", sum(all_samples) / len(all_samples))
```

```python
def check_cholesterol(level):
  if level < 200:
    return "normal"
  elif 200 <= level <= 239:
    return "Borderline High"
  else:
    return "High"
```

```python
cholesterol_list = []
while True:
  level = float(input("Enter Cholesterol Level (mg/dL) [or 0 to finish]: "))
  if level == 0:
    break
  status = check_cholesterol(level)
  print(f"Cholesterol Level: {level} mg/dL -> Status: {status}")
  cholesterol_list.append(level)
  print("\n--- LAB BATCH SUMMARY REPORT ---")
  print("Total Samples Processed:", len(cholesterol_list))
  print("Highest Cholesterol Level:", max(cholesterol_list))
  print("Lowest Cholesterol Level:", min(cholesterol_list))
  print("Average Cholesterol Level:", sum(cholesterol_list) / len(cholesterol_list))
```
