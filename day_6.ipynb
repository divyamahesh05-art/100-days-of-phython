```python
patients_data =[{"name": "Anil", "hba1c": 5.4},{"name": "Divya", "hba1c": 6.8},{"name": "Suresh", "hba1c": 7.2},{"name":"mahesh","hba1c":9.6}]
for patient in patients_data:
  if patient["hba1c"] > 6.5:
        patient["status"] = "Diabetic"
  else:
        patient["status"] = "Normal"
  print("--- ALL PATIENTS REPORT ---")
  print(f"Name: {patient['name']} | HbA1c: {patient['hba1c']} | Status: {patient['status']}")
```

```python
lab_samples=[{"id": "S1", "name": "Ramu", "hb": 11.5},{"id": "S2", "name": "Priya", "hb": 13.5},{"id": "S3", "name": "John", "hb": 10.2}]
for patient in lab_samples:
  if patient["hb"] < 12.5:
        patient["result"] = "anemic"
  else:
        patient["result"] = "Normal"
  print("--- ALL PATIENTS REPORT ---")
  print(f"patient:{patient['name']} | HbA1c: {patient['hb']} | Status: {patient['result']}")
```

```python
lab_records=[{"id": "P101", "name": "Sarah", "hb": 11.2, "fbs": 135},

{"id": "P102", "name": "Raj", "hb": 14.5, "fbs": 95},

{"id": "P103", "name": "John", "hb": 10.5, "fbs": 160}]
for patient in lab_records:
  if patient["hb"] < 12.0:
        patient["hb_status"] = "Low"
  else:
        patient["hb_status"] = "Normal"
  if patient["fbs"] > 126:
        patient["fbs_status"] = "High"
  else:
        patient["fbs_status"] = "Normal"
  if patient["hb_status"] !=  "Normal" or  patient["fbs_status"]!= "Normal":
    patient["flag"] = "ATTENTION REQUIRED"
  else:
    patient["flag"] = "Normal"
  print(patient)
```

```python
lab_records = []
print("--- ENTER PATIENT LAB DATA (Type 'exit' in name to finish) ---")
while True:
    name = input("Enter Patient Name: ")
    if name.lower() == "exit":
        break

    hb = float(input("Enter Hemoglobin (g/dL): "))
    fbs = float(input("Enter Fasting Blood Sugar (mg/dL): "))
    patient = {
        "name": name,
        "hb": hb,
        "fbs": fbs
    }
    lab_records.append(patient)
    print("Patient added successfully!\n")

```
