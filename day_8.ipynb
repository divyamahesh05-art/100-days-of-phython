```python
import pandas as pd
```

```python
data = {
    "Patient_ID": ["P101", "P102", "P103", "P104"],
    "Name": ["Sarah", "Raj", "John", "Meera"],
    "Age": [32, 65, 45, 70],
    "Hb_Level": [11.2, 14.5, 10.5, 13.0],
    "FBS": [135, 95, 160, 110]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("--- PATIENT LAB DATAFRAME ---")
print(df)
```

```python
low_hb_patients = df[df["Hb_Level"]<12.0]
print("--- LOW HEMOGLOBIN PATIENTS ---")
print(low_hb_patients)
```

```python
high_fbs_patients = df[df["FBS"]>140]
print("--- HIGH FBS PATIENTS ---")
print(high_fbs_patients)
```

```python
critical_patients = df[(df["Hb_Level"] < 12.0) & (df["FBS"] > 126)]

print("\n--- CRITICAL PATIENTS (BOTH LOW HB & HIGH FBS) ---")
print(critical_patients)
```

```python
avg_hb=df["Hb_Level"].mean()
avg_fbs=df["FBS"].mean()
max_fbs = df["FBS"].max()
min_fbs=df["FBS"].min()
min_hb=df["Hb_Level"].min()
print("--- LAB STATISTICS ---")
print("Average Hemoglobin:", avg_hb)
print("Average Blood Sugar:", avg_fbs)
print("Highest Blood Sugar:", max_fbs)
print("Lowest Hemoglobin:", min_hb)
```
